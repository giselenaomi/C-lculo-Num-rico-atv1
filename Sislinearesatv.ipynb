```python
import numpy as np
```

```python
# Exercício 1(a)


# Matriz dos coeficientes
A = np.array([
    [2, 1],
    [1, -3]
])

# Vetor dos termos independentes
b = np.array([7, -5])

# np.linalg.solve() resolve o sistema Ax = b
solucao = np.linalg.solve(A, b)

# Separando as soluções
x = solucao[0]
y = solucao[1]

# Mostrando os resultados
print("Matriz A:")
print(A)

print("\nVetor b:")
print(b)

print("\nSolução do sistema:")
print(f"x = {x}")
print(f"y = {y}")
```

```python
# Exercício 1(b)

import numpy as np

# Matriz dos coeficientes
A = np.array([
    [1, 2, -1],
    [3, -1, 2],
    [2, 1, 1]
])

# Vetor dos termos independentes
b = np.array([2, 9, 7])

# Resolve o sistema linear Ax = b
solucao = np.linalg.solve(A, b)

# Separando cada incógnita
x = solucao[0]
y = solucao[1]
z = solucao[2]

# Mostrando os resultados
print("Matriz A:")
print(A)

print("\nVetor b:")
print(b)

print("\nSolução do sistema:")
print(f"x = {x}")
print(f"y = {y}")
print(f"z = {z}")
```

```python
# EXERCÍCIO 2 - MÉTODO DA ELIMINAÇÃO DE GAUSS

# 1. Criando a matriz aumentada

M = np.array([
    [2.0,  1.0, -1.0,  8.0],
    [-3.0, -1.0,  2.0, -11.0],
    [-2.0,  1.0,  2.0, -3.0]
])

print("Matriz aumentada inicial:")
print(M)

# 2. Eliminação do x da segunda linha

# L2 <- L2 + (3/2)L1

M[1] = M[1] + (3/2) * M[0]

print("\nDepois de L2 <- L2 + (3/2)L1:")
print(M)



# 3. Eliminação do x da terceira linha

# L3 <- L3 + L1

M[2] = M[2] + M[0]

print("\nDepois de L3 <- L3 + L1:")
print(M)


# 4. Eliminação do y da terceira linha

# L3 <- L3 - 4L2

M[2] = M[2] - 4 * M[1]

print("\nDepois de L3 <- L3 - 4L2:")
print(M)


# 5. Substituição regressiva

# Primeiro encontramos z
z = M[2, 3] / M[2, 2]

# Depois encontramos y
y = (M[1, 3] - M[1, 2] * z) / M[1, 1]

# Finalmente encontramos x
x = (M[0, 3] - M[0, 1] * y - M[0, 2] * z) / M[0, 0]


# 6. Mostrando a solução


print("\nSolução do sistema:")
print(f"x = {x}")
print(f"y = {y}")
print(f"z = {z}")
```

```python

# EXERCÍCIO 3 - NÚMERO DE SOLUÇÕES

# ITEM (a)

M_a = np.array([
    [1.0, 1.0, 3.0],
    [2.0, 2.0, 6.0]
])

print("ITEM (a)")
print("Matriz aumentada inicial:")
print(M_a)

# L2 <- L2 - 2L1
M_a[1] = M_a[1] - 2 * M_a[0]

print("\nDepois de L2 <- L2 - 2L1:")
print(M_a)

print("\nConclusão:")
print("A segunda linha é 0 = 0.")
print("Portanto, o sistema possui infinitas soluções.")


# ITEM (b)

M_b = np.array([
    [1.0, 1.0, 3.0],
    [2.0, 2.0, 7.0]
])

print("\n\nITEM (b)")
print("Matriz aumentada inicial:")
print(M_b)

# L2 <- L2 - 2L1
M_b[1] = M_b[1] - 2 * M_b[0]

print("\nDepois de L2 <- L2 - 2L1:")
print(M_b)

print("\nConclusão:")
print("A segunda linha representa 0 = 1.")
print("Isso é uma contradição.")
print("Portanto, o sistema não possui solução.")


# ITEM (c)

M_c = np.array([
    [1.0, 1.0, 3.0],
    [2.0, -1.0, 0.0]
])

print("\n\nITEM (c)")
print("Matriz aumentada inicial:")
print(M_c)

# L2 <- L2 - 2L1
M_c[1] = M_c[1] - 2 * M_c[0]

print("\nDepois de L2 <- L2 - 2L1:")
print(M_c)

# Da segunda linha:
# -3y = -6
y = M_c[1, 2] / M_c[1, 1]

# Da primeira linha:
# x + y = 3
x = M_c[0, 2] - M_c[0, 1] * y

print("\nSolução:")
print(f"x = {x}")
print(f"y = {y}")

print("\nConclusão:")
print("O sistema possui uma única solução.")
```

```python

# MÉTODO DA ELIMINAÇÃO DE GAUSS SEM PIVOTAMENTO

def eliminacao_gauss(A, b):
    """
    Resolve o sistema linear Ax = b usando
    eliminação de Gauss sem pivotamento.

    Parâmetros:
        A -> matriz dos coeficientes
        b -> vetor dos termos independentes

    Retorna:
        x -> vetor solução
    """

    # 1. Fazemos cópias para não modificar A e b originais

    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)

    # Número de equações
    n = len(b)

    # 2. Formamos a matriz aumentada [A | b]

    M = np.column_stack((A, b))

    print("Matriz aumentada inicial:")
    print(M)


    # 3. ELIMINAÇÃO DE GAUSS


    for k in range(n - 1):

        # O elemento M[k,k] é o pivô atual
        pivo = M[k, k]

        # Como o método é sem pivotamento,
        # não vamos trocar linhas.
        if abs(pivo) < 1e-12:
            raise ValueError(
                "Pivô nulo ou muito próximo de zero. "
                "O método sem pivotamento não pode continuar."
            )

        # Eliminar os elementos abaixo do pivô


        for i in range(k + 1, n):

            # Multiplicador usado para zerar M[i,k]
            m = M[i, k] / pivo

            # Operação elementar:
            # Li <- Li - m * Lk
            M[i, k:] = M[i, k:] - m * M[k, k:]

        # Mostra a matriz após cada etapa
        print(f"\nApós eliminar usando o pivô da coluna {k + 1}:")
        print(M)


    # 4. SUBSTITUIÇÃO REGRESSIVA

    x = np.zeros(n)

    # Começamos pela última equação
    for i in range(n - 1, -1, -1):

        # Soma dos termos já conhecidos
        soma = 0

        for j in range(i + 1, n):
            soma += M[i, j] * x[j]

        # Isolando xi:
        #
        # M[i,i] * xi + soma = M[i,n]
        #
        # xi = (M[i,n] - soma) / M[i,i]

        x[i] = (M[i, n] - soma) / M[i, i]

    # 5. Retornamos a solução

    return x
```

```python
# TESTE COM O SISTEMA DO EXERCÍCIO 2

A = np.array([
    [2,  1, -1],
    [-3, -1,  2],
    [-2,  1,  2]
], dtype=float)

b = np.array([8, -11, -3], dtype=float)


# Executando nosso método de Gauss
solucao_gauss = eliminacao_gauss(A, b)

print("\n===================================")
print("SOLUÇÃO PELO MÉTODO DE GAUSS")
print("===================================")

print(f"x = {solucao_gauss[0]}")
print(f"y = {solucao_gauss[1]}")
print(f"z = {solucao_gauss[2]}")
```

```python
# COMPARAÇÃO COM NUMPY

solucao_numpy = np.linalg.solve(A, b)

print("Solução pelo nossa método de Gauss:")
print(solucao_gauss)

print("\nSolução usando numpy.linalg.solve:")
print(solucao_numpy)

print("\nDiferença entre os resultados:")
print(solucao_gauss - solucao_numpy)
```

```python
#Interpretação: O algoritmo de eliminação de Gauss implementado encontrou a solução \(x=2\), \(y=3\) e \(z=-1\). O resultado foi igual ao obtido pela função numpy.linalg.solve(), indicando que a implementação apresentou o resultado esperado para o sistema testado. A eliminação de Gauss transforma a matriz aumentada em uma matriz triangular superior e, posteriormente, utiliza a substituição regressiva para encontrar as incógnitas.
```

```python
# EXERCÍCIO 5 - MODELO DE BALANÇO TÉRMICO

# Matriz dos coeficientes
A = np.array([
    [4, -1,  0],
    [-1, 4, -1],
    [0, -1,  3]
], dtype=float)

# Vetor dos termos independentes
b = np.array([15, 10, 10], dtype=float)


# Resolvendo pelo nosso método de Gauss

T = eliminacao_gauss(A, b)

# Mostrando as temperaturas calculadas

print("\n===================================")
print("RESULTADO DO MODELO TÉRMICO")
print("===================================")

print(f"T1 = {T[0]}")
print(f"T2 = {T[1]}")
print(f"T3 = {T[2]}")
```
